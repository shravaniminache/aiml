```python
import random

class Environment:

    def __init__(self):
        self.locationCondition = {
            'A': random.randint(0, 1),
            'B': random.randint(0, 1)
        }

    def display_environment(self):
        print("\nCurrent Environment:")
        for location, condition in self.locationCondition.items():
            status = "Dirty" if condition == 1 else "Clean"
            print(f"Location {location}: {status}")

class SimpleReflexVacuumAgent:

    def __init__(self, environment):

        self.environment = environment
        self.vacuumLocation = random.choice(['A', 'B'])
        self.score = 0
        self.cleaning_actions = 0
        self.movement_actions = 0
        self.battery = 10
        self.steps = 0

        print("\n====================================")
        print("     SIMPLE REFLEX VACUUM AGENT")
        print("====================================")

        print(f"Vacuum is initially placed at Location {self.vacuumLocation}")
        print(f"Initial Battery: {self.battery}")

  
    def perceive(self):

        location = self.vacuumLocation
        condition = self.environment.locationCondition[location]

        return location, condition

 

    def choose_action(self, location, condition):
        if condition == 1:
            return "SUCK"
        elif location == 'A':
            return "MOVE_RIGHT"

        else:
            return "MOVE_LEFT"

   

    def perform_action(self, action):

 
        if self.battery <= 0:
            print("\nBattery exhausted!")
            return False

       

        if action == "SUCK":

            print(f"Location {self.vacuumLocation} is Dirty.")
            print("Action: Sucking dirt...")

            self.environment.locationCondition[
                self.vacuumLocation
            ] = 0

            self.cleaning_actions += 1

            self.battery -= 1

 
            self.score += 1

            print(f"Location {self.vacuumLocation} has been Cleaned.")

        elif action == "MOVE_RIGHT":

            print("Current location is Clean.")
            print("Action: Moving Right → Location B")

            self.vacuumLocation = 'B'

            self.movement_actions += 1

            self.battery -= 1

          
            self.score -= 1

        elif action == "MOVE_LEFT":
            print("Current location is Clean.")
            print("Action: Moving Left → Location A")
            self.vacuumLocation = 'A'
            self.movement_actions += 1
            self.battery -= 1
            self.score -= 1
        return True


    def is_environment_clean(self):

        return (
            self.environment.locationCondition['A'] == 0
            and
            self.environment.locationCondition['B'] == 0
        )
    def generate_dirt(self):

        
        random_location = random.choice(['A', 'B'])

      
        if random.random() < 0.30:

            self.environment.locationCondition[random_location] = 1

            print(
                f"\nNew dirt appeared at Location {random_location}!"
            )
    def run(self, max_steps=10):

        print("\nStarting Vacuum Cleaner...\n")

        for step in range(max_steps):

            self.steps += 1

            print("\n------------------------------------")
            print(f"STEP {self.steps}")
            print("------------------------------------")

            self.environment.display_environment()

            print(f"\nVacuum Location: {self.vacuumLocation}")
            print(f"Battery Remaining: {self.battery}")

    
            location, condition = self.perceive()

            action = self.choose_action(location, condition)

            print(f"Selected Action: {action}")

            success = self.perform_action(action)

            if not success:
                break

            self.generate_dirt()

        self.display_performance()


    def display_performance(self):

        print("\n====================================")
        print("       PERFORMANCE MEASUREMENT")
        print("====================================")

        self.environment.display_environment()

        print("\nPerformance Statistics:")
        print(f"Total Steps        : {self.steps}")
        print(f"Cleaning Actions   : {self.cleaning_actions}")
        print(f"Movement Actions   : {self.movement_actions}")
        print(f"Battery Remaining  : {self.battery}")
        print(f"Final Score        : {self.score}")

        if self.is_environment_clean():
            print("\nResult: Environment is CLEAN.")
        else:
            print("\nResult: Some locations are still DIRTY.")

theEnvironment = Environment()

theEnvironment.display_environment()

theVacuum = SimpleReflexVacuumAgent(theEnvironment)

theVacuum.run(max_steps=10)
```
